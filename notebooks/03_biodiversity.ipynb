# 03 — Biodiversity profile
Move beyond tree counts to taxonomic diversity and concentration.

**Illustrative example:** The counts below are an illustrative concentration example, not measured historical results. See 05 for verified historical checkpoints.

```python
from collections import Counter

genera = Counter({'Platanus':42000,'Aesculus':18000,'Tilia':16000,'Other':90000})
total = sum(genera.values())
shares = {name:round(count/total,3) for name,count in genera.items()}
concentration = round(sum(share**2 for share in shares.values()),3)
shares, concentration
```

Taxonomic diversity is not ecological health by itself. Interpret it with age, site conditions, maintenance and resilience expertise.
